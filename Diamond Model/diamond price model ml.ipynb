```python
import pandas as pd
import joblib 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

df = pd.read_csv('diamonds.csv')

# Data Cleaning
print("\nMissing Values:")
print(df.isnull().sum())

df = df.drop_duplicates()

print("\nDuplicates Removed!")

num_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
X = df.drop('price',axis=1)
Y = df['price']

X = pd.get_dummies(X)

encoded_cols = X.columns.to_list()

numeric_cols = ["carat","depth","table","x","y","z"]

Scaler = StandardScaler()
X[numeric_cols] = Scaler.fit_transform(X[numeric_cols])

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.33,
    random_state=42
)

# Train Model
model = LinearRegression()
model.fit(X_train, Y_train)

from sklearn.metrics import r2_score

# Make predictions
y_pred = model.predict(X_test)

# Calculate R² Score
r2 = r2_score(Y_test, y_pred)

print("R² Score:", r2)

print("First 10 Predicted Values:")
print(y_pred[:10])
print("\nFirst 10 Actual Values:")
print(Y_test.iloc[:10].values)

Comparison = pd.DataFrame({
    "Actual Values:": Y_test.iloc[:10].values,
    "Predicted Values": y_pred[:10]
})

print("\nComparison of Actual vs Predicted Values:")
print(Comparison)
```

```python
# Save Files
joblib.dump(model, "LR_Diamond_Price.pkl")
joblib.dump(Scaler, "scaler_Diamond_Price.pkl")
joblib.dump(encoded_cols, "columns_Diamond_Price.pkl")
print("Saved Successfully")
```
