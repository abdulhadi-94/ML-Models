```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
import joblib

# Load dataset
df = pd.read_csv("Laptop_price - dataset.csv")

# Input and Output
X = df.drop("Price (Euro)", axis=1)
y = df["Price (Euro)"]

# One-Hot Encode
X = pd.get_dummies(X)

# Save feature names
encoded_columns = X.columns.tolist()

# Scale only numeric columns
numeric_cols = X.select_dtypes(include=['int64', 'float64']).columns

scaler = StandardScaler()
X[numeric_cols] = scaler.fit_transform(X[numeric_cols])

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.33,
    random_state=42
)
```

```python
# ===========================================
# Q3: Regression Algorithms
# ===========================================

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

import pandas as pd
import numpy as np

# Dictionary of Regression Models
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(random_state=42),
    "Support Vector Regressor": SVR(),
    "K-Nearest Neighbors Regressor": KNeighborsRegressor()
}

# Store Results
results = []

# ===========================================
# Train and Evaluate
# ===========================================

for name, model in models.items():

    print("="*60)
    print(name)
    print("="*60)

    # Train
    model.fit(X_train, y_train)

    # Predict
    y_pred = model.predict(X_test)

    # Evaluation Metrics
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)

    print(f"R² Score : {r2:.4f}")
    print(f"MAE      : {mae:.2f}")
    print(f"MSE      : {mse:.2f}")
    print(f"RMSE     : {rmse:.2f}")

    # Save Results
    results.append({
        "Model": name,
        "R² Score": r2,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse
    })

# ===========================================
# Comparison Table
# ===========================================

comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by="R² Score",
    ascending=False
)

print("\n")
print("="*60)
print("MODEL COMPARISON TABLE")
print("="*60)

print(comparison_df)

# ===========================================
# Best Model
# ===========================================

best_model = comparison_df.iloc[0]

print("\nBest Regression Model:")
print(best_model)
```

```python
# ===========================================
# Best Regression Model
# ===========================================

from sklearn.metrics import r2_score

regression_models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(random_state=42),
    "Support Vector Regressor": SVR(),
    "KNN Regressor": KNeighborsRegressor()
}

best_r2 = -100
best_regressor = None
best_regressor_name = ""

for name, model in regression_models.items():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    r2 = r2_score(y_test, y_pred)

    if r2 > best_r2:
        best_r2 = r2
        best_regressor = model
        best_regressor_name = name

print("="*50)
print("Best Regression Model")
print("="*50)
print(best_regressor_name)
print("R² Score :", round(best_r2,4))


# -------------------------------------------------
# Save Regression Model
# -------------------------------------------------

joblib.dump(best_regressor, "best_regressor.pkl")
joblib.dump(scaler, "regression_scaler.pkl")
joblib.dump(encoded_columns, "regression_columns.pkl")

print("\nRegression model saved successfully!")
```
