```python
# Q1. Manual Hyperparameter Tuning – KNN

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# Load Dataset
df = pd.read_csv("Iris.csv")

# Display first 5 rows
print(df.head())

# Features and Target
X = df.drop(columns=["Species", "Id"])
y = df["Species"]

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.33,
    random_state=42
)

# Different values of K
neighbors = [3, 5, 7, 11, 13, 15]

best_score = 0
best_k = None

print("KNN Test Accuracies\n")

for k in neighbors:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)

    score = model.score(X_test, y_test)

    print(f"n_neighbors = {k} --> Accuracy = {score:.4f}")

    if score > best_score:
        best_score = score
        best_k = k

print("\nBest Value of K:", best_k)
print("Highest Accuracy:", round(best_score,4))
```

```python
# Q.2.Hyper Tuning

from sklearn.svm import SVC

# Change these values manually
C_value = 10         
kernel_type = "rbf"   

# Create SVM model
model = SVC(C=C_value, kernel=kernel_type)

# Train the model
model.fit(X_train, y_train)

# Test Accuracy
score = model.score(X_test, y_test)

print("SVM Results")
print("C Value :", C_value)
print("Kernel  :", kernel_type)
print("Accuracy:", round(score, 4))

# Change these values manually
C_value = 10         
kernel_type = "linear"  

# Create SVM model
model = SVC(C=C_value, kernel=kernel_type)

# Train the model
model.fit(X_train, y_train)

# Test Accuracy
score = model.score(X_test, y_test)

print("\nSVM Results")
print("C Value :", C_value)
print("Kernel  :", kernel_type)
print("Accuracy:", round(score, 4))
```

```python
# Q3. Grid Search CV

from sklearn.model_selection import GridSearchCV

# Parameter Grid
param_grid = {
    "C": [1, 10, 20],
    "kernel": ["linear", "rbf"]
}

# Grid Search
grid = GridSearchCV(
    estimator=SVC(),
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)

# Train
grid.fit(X_train, y_train)

# Results DataFrame
results = pd.DataFrame(grid.cv_results_)

print(results[["param_C", "param_kernel", "mean_test_score"]])

print("\nBest Parameters:")
print(grid.best_params_)

print("\nBest Cross Validation Score:")
print(round(grid.best_score_,4))
```

```python
# Q4. Randomized Search CV

from sklearn.model_selection import RandomizedSearchCV

# Parameter Grid
param_dist = {
    "C": [1, 10, 20],
    "kernel": ["linear", "rbf"]
}

# Randomized Search
random_search = RandomizedSearchCV(
    estimator=SVC(),
    param_distributions=param_dist,
    n_iter=5,
    cv=5,
    random_state=42,
    scoring="accuracy"
)

# Train
random_search.fit(X_train, y_train)

# Results DataFrame
random_results = pd.DataFrame(random_search.cv_results_)

print(random_results[
    ["param_C", "param_kernel", "mean_test_score", "rank_test_score"]
])

print("\nBest Parameters:")
print(random_search.best_params_)

print("\nBest Cross Validation Score:")
print(round(random_search.best_score_,4))

print("\nComparison with Grid Search")

print("Grid Search Best Parameters :", grid.best_params_)
print("Grid Search Best Score      :", round(grid.best_score_,4))

print("Random Search Best Parameters :", random_search.best_params_)
print("Random Search Best Score      :", round(random_search.best_score_,4))
```
